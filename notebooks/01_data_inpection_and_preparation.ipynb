```python
missing_data = df[df["value"].isnull()]

missing_summary = (
    missing_data
    .groupby(["country", "indicator", "indicator_code", "year"])
    .size()
    .reset_index(name="missing_count")
)

print(missing_summary)
```

```python
indicators = df[
    ["indicator", "indicator_code"]
].drop_duplicates()

print(indicators.to_string(index=False))
```

```python
indicator_summary = (
    df.groupby(["indicator", "indicator_code"])["value"]
    .agg(["count", "min", "max", "mean"])
    .reset_index()
)

print(indicator_summary)
```

```python
print("Countries:")
print(df["country"].unique())

print("\nNumber of countries:")
print(df["country"].nunique())

print("\nIndicators:")
print(df["indicator"].unique())
print(df["indicator"].nunique())

print("\nYear range:")
print(df["year"].min(), "-", df["year"].max())
```

```python
country_summary = (
    df.groupby(["country", "indicator"])["value"]
    .agg(["count", "min", "max", "mean"])
    .reset_index()
)

print(country_summary.to_string(index=False))
```

```python
missing_by_country = (
    df[df["value"].isnull()]
    .groupby(["country", "indicator", "year"])
    .size()
    .reset_index(name="missing_count")
)

print(missing_by_country.to_string(index=False))
```

```python
indicator_units = {
    "inflation": "Percentage (%)",
    "gdp_growth": "Percentage (%)",
    "gdp_per_capita": "Current US Dollars (USD)",
    "unemployment": "Percentage of total labour force (%)",
    "exchange_rate": "Local currency units per USD",
    "population": "Number of people",
    "internet_users": "Percentage of population (%)",
    "oil_rents": "Percentage of GDP (%)"
}

df["unit"] = df["indicator"].map(indicator_units)

df.head()
```

```python
duplicate_records = df.duplicated(
    subset=["country_code", "indicator_code", "year"]
).sum()

print("Duplicate country-indicator-year records:", duplicate_records)
```

```python
df = df.sort_values(
    by=["country", "indicator", "year"],
    ascending=[True, True, True]
).reset_index(drop=True)
```
